```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset=pd.read_csv("/home/jobayer/Git/titanic/train.csv")
```

```python
dataset.info()
```

```python
d_cr=dataset.select_dtypes("int64","float64").corr()
```

```python
d_cv=dataset.select_dtypes("int64","float64").cov()
```

```python
plt.figure(figsize=(4,3))
sns.heatmap(d_cr,annot=True)
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.heatmap(d_cv,annot=True)
plt.show()
```
