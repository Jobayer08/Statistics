```python
import scipy.stats as stats
import numpy as np
```

```python
stats.norm.ppf(0.95)
```

```python
s_x=90
p_std=20
p_u=82
n=81
sig=0.05
```

```python
z_c=(s_x-p_u)/(p_std/(np.sqrt(n)))
```

```python
z_ta=stats.norm.ppf(1-sig)
```

```python
if z_c>z_ta:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```
