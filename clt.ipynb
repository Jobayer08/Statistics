```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
p_d=[np.random.randint(10,100) for i in range(1000)]
p_t=pd.DataFrame({"p_d":p_d})
p_t
```

```python
plt.figure(figsize=(3,2))
sns.kdeplot(x='p_d',data=p_t)
plt.show()
```

```python
s_m=[]
for n_s in range(50):
    s_d=[]
    for d in range(500):
        s_d.append(np.random.choice(p_d))
    s_m.append(np.mean(s_d))    
```

```python
s_M=pd.DataFrame({"s_m":s_m})
```

```python
plt.figure(figsize=(3,2))
sns.kdeplot(x="s_m",data=s_M)
plt.show()
```
